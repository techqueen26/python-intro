```python
inventory = [
    ("rice", 3000, 30),
    ("beans", 2800, 20),
    ("flour", 2200, 15),
    ("bread", 1400, 11),

]
inventory.append(
    ("orange", 2000, 5)
)
inventory.append(
    ("grapes", 2200, 4)
)
print(inventory)

#remove selected products by unpacking
for product in inventory:
    name, price, qty =product
    if product[0] =="beans":
        inventory.remove(product)

print(inventory)

#print enumerated inventory

for num, product in enumerate(inventory, start=1):
    print(f"{num} -{product}")

    total_price = 0

    for product in inventory:
        name, price, qty = product
        total_price += product[1] * product[2]

    print(f"total inventory price = {total_price}")

    # get most expensive product

    most_expensive = inventory[3]

    for product in inventory:
        if product[1] > most_expensive[1]:
            most_expensive = product

    print(most_expensive[0:-1])
```

```python
#create a dictionary mapping course names to a set od enrolled students
data_set = {
    "pes": {"sola", "seun", "seyi"},
    "com": {"dad", "sam", "bob"}
}
data_set["pes"].add("kola")
data_set["pes"].add("bola")

#get intersection and difference of course a and b
student_in_both = data_set["pes"].intersection(data_set["com"])
only_pes = data_set["pes"].difference(data_set["com"])

print(data_set)
print(student_in_both)
print(only_pes)

#count student per course

print("=" * 20)
print("solution  4")

print("=" * 20)

for course, students in data_set.items():
    print(f"{course} - {len(students)}")

# find the course with the highest student count
highest_student = 0
most_student = ""

for course, students in data_set.items():
    if len(students) > highest_student:
        highest_student = len(students)
        most_student = course

print(most_student)



   
         
```
